## Filtramos por Portabicibletas

```python
# Lista de códigos a filtrar (portabicibletas)
codigos = [
    "952-295", "952-432", "952-084", "952-086", "952-087", "940-506AU", "940-511AU", "940-515AU", "952-019", "952-022", "952-023", "952-026", "952-028", "952-051", "952-052", "940-001", "940-005", "940-010", "940-506", "940-508", "940-511", "940-515", "940-520", "940-525", "940-006", "940-007", "940-544", "940-563", "940-564", "940-565", "940-012", "940-015", "940-016", "940-541", "940-560", "940-561", "940-562", "R130101001"]
# Filtramos el DataFrame
df_AlbaranesFiltradoML_filtrado1 = df_AlbaranesFiltradoML[df_AlbaranesFiltradoML["CodigoArticulo"].isin(codigos)]
```

```python
len(df_AlbaranesFiltradoML_filtrado1)
```

## Filtramos por años a partir de 2021

```python
# Convertimos la columna a datetime (asumiendo formato día/mes/año)
df_AlbaranesFiltradoML_filtrado1["FechaAlbaran"] = pd.to_datetime(df_AlbaranesFiltradoML_filtrado1["FechaAlbaran"], dayfirst=True)

# Definimos la fecha límite
fecha_limite = pd.to_datetime("31/12/2020", dayfirst=True)

# Filtramos el DataFrame
df_AlbaranesFiltradoML_filtrado2 = df_AlbaranesFiltradoML_filtrado1[df_AlbaranesFiltradoML_filtrado1["FechaAlbaran"] > fecha_limite]

```

```python
len(df_AlbaranesFiltradoML_filtrado2)
```

### Analizamos cuantas lineas por tipo de canal tenemos para ver si Usuario Final es relevante en nuestro anlálisis o no

```python
tabla_recuento = df_AlbaranesFiltradoML_filtrado2.groupby("CodigoCanal").size().reset_index(name="Recuento")
tabla_recuento
```
