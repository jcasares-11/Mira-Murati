## Datos externos - eventos ciclismo

Real Federacion Española de Turismo: https://rfec.com/index.php/es/smartweb/seccion/calendario/rfec/2022

Periodo: 2021-2024

La idea es que en el input final del modelo haya una columna que se llame 'evento' y haya un 0 si no hay evento y un 1 si si hay evento

```python
import pandas as pd
```

```python
eventos21 = pd.read_excel('Calendario Ciclismo 2021.xlsx')
eventos22 = pd.read_excel('Calendario Ciclismo 2022.xlsx')
eventos23 = pd.read_excel('Calendario Ciclismo 2023.xlsx')
eventos24 = pd.read_excel('Calendario Ciclismo 2024.xlsx')
```

```python
eventos = pd.concat([eventos21, eventos22, eventos23, eventos24], ignore_index=True)
eventos
```

```python
eventos_ciclismo = eventos.drop(columns = ['Fecha fin', 'Modalidad','Clase','Prueba', 'Lugar','Provincia','Club','Email', 'Info', 'Categorías','Observaciones'])
eventos_ciclismo
```

```python
eventos_ciclismo['Evento'] = 1
```

```python
eventos_ciclismo
```

```python
eventos_ciclismo.groupby(['Fecha inicio']).agg({'Evento':sum}).reset_index()
```

*Nota: en el input, en las fechas que no haya evento - poner un 0

```python
eventos_ciclismo.dtypes
```

```python
eventos_ciclismo['Fecha inicio'] = pd.to_datetime(eventos_ciclismo['Fecha inicio'], errors='coerce')
```
